# LibreYOLO

```python
from libreyolo import LibreYOLO

model = LibreYOLO("LibreYOLO9c.pt")

results = model.train(
    data=r"../../data/license_plate_detection/data.yaml",
    
    # Schedule
    epochs=250,
    batch=32,
    imgsz=640,
    
    # Optimizer
    lr0=0.01,
    optimizer="SGD",
    
    # System
    device="cuda",
    workers=8,
    seed=42,
    
    # Output
    project="runs/train",
    name="yolo9_exp",
    exist_ok=True,
    
    # Training features
    amp=True,
    patience=50,
    resume=False,
)

print(f"Best mAP50-95: {results["best_map50_95"]:.3f}")
print(f"Best checkpoint: {results["best_checkpoint"]}")
```
