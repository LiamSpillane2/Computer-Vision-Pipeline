```python

import json
import os
from pathlib import Path
from PIL import Image
import pandas as pd

main_dir = Path(r"C:\Users\Installer\Programming\python_assessment\Computer-Vision-Pipeline\data\formatted")
output_file = "metadata.json"

metadata = []
columns = []

for root, dirs, files in os.walk(main_dir):
    for filename in files:
        full_path = os.path.join(root, filename)
        name = filename.split(".")[0]
        file_type = filename.split("_")[2].split(".")[1]
        size = os.path.getsize(full_path)
        width = None
        height = None
        class_ = None
        row_count = None

        if file_type == "jpg":
            with Image.open(full_path) as img:
                width, height = img.size
        else:
            with open(full_path, "r") as file:
                rows = file.readlines()
                row_count = len(rows)
                file.close()
                


        metadata.append({
            "Name": name,
            "type": file_type,
            'width': width,
            'height': height,
            "size": size,
            "path": full_path,
            "classes": row_count
        })

df = pd.DataFrame(metadata)
df.head()

# with open("metadata.json", "w") as f:
#     json.dump(metadata, f, indent=4)
```
