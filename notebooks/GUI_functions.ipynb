```python
# fuction to add extracted text, text probability, and zero shot classifier labels and probabilities
def add2metadata(model_data, ocr_output):

    model_df = pd.DataFrame.from_dict(model_data)
    ocr_df = pd.DataFrame.from_dict(ocr_output)
    cols_to_add = ['text', 'confidence', "bounding_box", "file_name" ]

    # zsc_df = pandas.read_json(zsc_output)
    # model_df = pandas.read_json(model_output)
    model_df = model_df.join(ocr_df[cols_to_add])

    return model_df

metadata = add2metadata(model_data, ocr_output)
```

```python
#Search Function
def search4id(metadata, search_text):
    cols = metadata.columns.to_list()
    for x in range(len(metadata)):
        if search_text == metadata['text'].iloc[x]:
            print("Match found!")
            for y in range(len(cols)):
                print(f"{cols[y]}: {metadata[cols[y]].iloc[x]}")
    
search4id(metadata, search_text='JZW-2689')
```
